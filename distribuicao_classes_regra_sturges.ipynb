```python
n = dados.shape[0]
k = int(round(1 + (10/3) * np.log10(n),0))
```

```python
labels = [f'Classe: {1 + i}' for i in range(k)]
```

```python
new_frequencia = pd.cut(
    x = dados.Renda,
    bins = k,
    labels = labels).value_counts()

new_frequencia
```

```python
frequencia_relativa = pd.cut(
    x = dados.Renda,
    bins = k,
    labels = labels).value_counts(normalize=True) * 100

frequencia_relativa
```

```python
pd.DataFrame({
    'Frequência Absoluta': new_frequencia,
    'Frequência Relativa': frequencia_relativa
})
```

```python
intervalos = pd.cut(x = dados.Renda, bins = k).cat.categories

pd.DataFrame({
    'Classe': [f'{i + 1}' for i in range(len(intervalos))],
    'Limite Inferior': [f'R$ {i.left}' for i in intervalos],
    'Limite Superior': [f'R$ {i.right}' for i in intervalos]
})
```
