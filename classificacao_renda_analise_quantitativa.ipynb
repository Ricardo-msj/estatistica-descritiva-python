```python
classes = [0, 1576,3152, 7880, 15760, 200000]
labels = ['E', 'D', 'C', 'B', 'A']
```

```python
frequencia_renda = pd.cut(
    x = dados.Renda,
    bins = classes,
    labels = labels,
    include_lowest = True
    ).value_counts()

frequencia_renda.sort_index(ascending=False)
```

```python
percentual_renda = pd.cut(
    x = dados.Renda,
    bins = classes,
    labels = labels,
    include_lowest = True
    ).value_counts(normalize=True) * 100

percentual_renda.sort_index(ascending=False)
```

```python
distri_freq_quantitativa = pd.DataFrame({'Frequencia': frequencia_renda, 'Percentual': percentual_renda}).sort_index(ascending=False)
distri_freq_quantitativa.rename_axis(None, axis='index', inplace = True)
distri_freq_quantitativa.rename_axis('Classes', axis='columns', inplace=True)
distri_freq_quantitativa

```
